**1a.** Write a query that selects `date`, `store`, and `revenue` from `sales` where `revenue` is greater than 4000. Limit to 10 rows.

## Part 2 — Aggregation (20%)


**2a.** Write a query that computes total `revenue` per `category`, sorted highest to lowest.


```python
# TODO 2a
query_2a = """
SELECT category, SUM(revenue) AS total_revenue
FROM sales
GROUP BY category
ORDER BY total_revenue DESC
"""
df_2a = pd.read_sql_query(query_2a, conn)
df_2a
```

**2b.** Write a query that computes average `units_sold` per `store`, but only shows stores where the average exceeds 50 (use `HAVING`).


```python
# TODO 2b
query_2b = """
SELECT store, AVG(units_sold) AS avg_units
FROM sales
GROUP BY store
HAVING AVG(units_sold) > 50
"""
df_2b = pd.read_sql_query(query_2b, conn)
df_2b
```

## Part 3 — Joins (25%)


**3a.** Write an `INNER JOIN` query between `sales` and `stores` that returns `store`, `category`, `revenue`, and `region` for the first 10 rows.


```python
# TODO 3a
query_3a = """
SELECT s.store, s.category, s.revenue, st.region
FROM sales AS s
INNER JOIN stores AS st ON s.store = st.store
LIMIT 10
"""
df_3a = pd.read_sql_query(query_3a, conn)
df_3a
```

**3b.** Write a `LEFT JOIN` query starting from `stores`, and print only the row(s) where there is no matching sale (hint: use `IS NULL`).


```python
# TODO 3b
query_3b = """
SELECT st.store, st.region
FROM stores AS st
LEFT JOIN sales AS s ON st.store = s.store
WHERE s.store IS NULL
"""
df_3b = pd.read_sql_query(query_3b, conn)
df_3b
```

**3c.** Write a query that joins `sales` and `stores`, then computes total `revenue` per `region`, sorted highest to lowest.


```python
# TODO 3c
query_3c = """
SELECT st.region, SUM(s.revenue) AS total_revenue
FROM sales AS s
INNER JOIN stores AS st ON s.store = st.store
GROUP BY st.region
ORDER BY total_revenue DESC
"""
df_3c = pd.read_sql_query(query_3c, conn)
df_3c
```

**5a.** In Part 3b, you found the store with no matching sales. Why does an `INNER JOIN` fail to show this store at all, while a `LEFT JOIN` does?

*Your answer here:*


An INNER JOIN requires a matching key in both tables, so any store (like Plattsburgh) without corresponding records in the sales table is completely dropped from the result. A LEFT JOIN retains all records from the left table (stores), filling in NULL for the missing columns from the right table (sales), which allows unmatched stores to appear.

**5b.** Why is a parameterized query (Part 4a) generally a better habit than building a query with an f-string, even for a class assignment where there's no real security risk?

Parameterized queries automatically escape input values, preventing SQL injection vulnerabilities and formatting errors (such as unescaped single quotes in text).
