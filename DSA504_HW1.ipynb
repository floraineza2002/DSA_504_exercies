## Part 3 — Sort (20%)


**3a.** Sort the full dataset by `revenue` in descending order, and print just the `date`, `store`, `category`, and `revenue` columns for the top 5 rows.


```python
# TODO 3a
sorted_3a = sales.sort_values("revenue", ascending=False)

print(sorted_3a[["date", "store", "category", "revenue"]].head(5))
```

**3b.** Sort the dataset by `store` (ascending) and then `units_sold` (descending) within each store. Print the first 10 rows of the result, showing only `store` and `units_sold`.


```python
# TODO 3b
sorted_3b = sales.sort_values(
    by=["store", "units_sold"],
    ascending=[True, False]

)

print(sorted_3b[["store", "units_sold"]].head(10))
```

## Part 4 — Group & Aggregate (30%)


**4a.** Compute total revenue per store. Which store has the highest total?


```python
# TODO 4a
revenue_per_store = sales.groupby("store")["revenue"].sum()

print(revenue_per_store)
print("\nStore with the highest total revenue:")
print(revenue_per_store.idxmax())
```

**4b.** Compute the average `units_sold` per category (using the raw, uncleaned category column — yes, this means you'll get more than 5 groups, and that's expected for this assignment).


```python
# TODO 4b
avg_units_sold_category = sales.groupby("category")["units_sold"].mean()

print(avg_units_sold_category)
```

**4c.** Using `.agg()`, compute both the **sum** and the **count** of `revenue` per store in a single call. Then sort the result so the store with the highest total revenue appears first.


```python
# TODO 4c
revenue_summary = sales.groupby("store")["revenue"].agg(["sum", "count"])

revenue_summary = revenue_summary.sort_values("sum", ascending=False)

print(revenue_summary)
```

**4d.** Group by **both** `store` and `category` together, and compute total revenue for each store/category combination. Print just the first 10 rows of the result.


```python
# TODO 4d
revenue_by_store_category = sales.groupby(

    ["store", "category"]

)["revenue"].sum()

print(revenue_by_store_category.head(10))
```
